```python
df_resumen = pd.read_csv(
    "cuadro16_resumen.csv",
    sep=","
)
df_resumen.head(10)
```

```python
df_resumen.dtypes
```

```python
print("La cantidad de filas en la base de datos es: ", len(df_resumen))
```

```python
print("La cantidad de columnas en la base de datos es: ", len(df_resumen.columns))
```

```python
print("La cantidad de filas y columnas en la base de datos es: ", df_resumen.shape)
```

```python
datos_nulos1=df_resumen.isna().sum()
print("cantidad de datos nulos es: ",datos_nulos1)
```

```python
total_nulos=df_resumen.isnull().sum().sum()
print("La cantidad total de datos nulos es ", total_nulos)
```

```python
datos_numericos=df_resumen.select_dtypes(include="number")
datos_numericos.head(10)
```

```python
datos_NO_numericos=df_resumen.select_dtypes(exclude="number")
datos_NO_numericos.head(10)
```

```python
media=np.mean(datos_numericos)
print("la media de los datos es: ", media)
```

```python
mediana=np.median(datos_numericos)
print("la mediana de los datos es ", mediana)
```

```python
varianza=np.var(datos_numericos)
np.set_printoptions(suppress=True)
print(f"la varianza de los datos es {varianza}")
```

```python
desv_std=np.std(datos_numericos)
print("la desviacion estandar de los datos es ", desv_std)
```

```python
minima=np.max(datos_numericos)
print("la minima de los datos es ", minima)
```

```python
maximo=np.max(datos_numericos)
print("el maximo de los datos es ",maximo)
```

```python
moda=df_resumen.mode()
print("la moda de los datos es\n ",moda)
```
