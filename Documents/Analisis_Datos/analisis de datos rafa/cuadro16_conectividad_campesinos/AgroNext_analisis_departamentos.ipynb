```python
# Carga y depuración del archivo departamental
df_dept = pd.read_csv("cuadro16_departamentos.csv", encoding="utf-8-sig")
df_dept_clean = df_dept.drop(index=26) # Exclusión de San Andrés (0 hogares)

# Cálculo de Medidas de Tendencia Central por Columna
columnas_num = ['Total_hogares_miles', 'Hogares_internet_miles', 'Porcentaje_internet', 'CVE_internet']

print("=== MEDIDAS DE TENDENCIA CENTRAL ===")
for col in columnas_num:
    media = df_dept_clean[col].mean()
    mediana = df_dept_clean[col].median()
    moda = df_dept_clean[col].mode().values
    
    print(f"\nColumna: {col}")
    print(f" -> Media: {media:.4f}")
    print(f" -> Mediana: {mediana:.4f}")
    print(f" -> Moda: {moda[0] if len(moda) == 1 else 'Multimodal (sin repetición dominante)'}")
```

```python
# Cálculo de Medidas de Dispersión por Columna
print("=== MEDIDAS DE DISPERSIÓN Y RANGOS ===")
for col in columnas_num:
    varianza = df_dept_clean[col].var()
    std_dev = df_dept_clean[col].std()
    minimo = df_dept_clean[col].min()
    maximo = df_dept_clean[col].max()
    rango = maximo - minimo
    
    print(f"\nColumna: {col}")
    print(f" -> Varianza (s^2): {varianza:.4f}")
    print(f" -> Desviación Estándar (s): {std_dev:.4f}")
    print(f" -> Mínimo: {minimo:.4f}")
    print(f" -> Máximo: {maximo:.4f}")
    print(f" -> Rango Absoluto: {rango:.4f}")
```

```python
# Cálculo de Cuartiles y Rango Intercuartílico (IQR)
print("=== CUARTILES Y RANGO INTERCUARTÍLICO (IQR) ===")
for col in columnas_num:
    q1 = df_dept_clean[col].quantile(0.25)
    q2 = df_dept_clean[col].quantile(0.50)
    q3 = df_dept_clean[col].quantile(0.75)
    iqr = q3 - q1
    
    print(f"\nColumna: {col}")
    print(f" -> Q1 (25%): {q1:.4f}")
    print(f" -> Q2 (50% / Mediana): {q2:.4f}")
    print(f" -> Q3 (75%): {q3:.4f}")
    print(f" -> IQR (Q3 - Q1): {iqr:.4f}")
```

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Carga y filtrado de datos departamentales (n = 32 departamentos válidos)
df_dept = pd.read_csv("cuadro16_departamentos.csv", encoding="utf-8-sig")
df_dept_clean = df_dept.drop(index=26).reset_index(drop=True)

# Definición de variables predictora (X) y respuesta (y)
X = df_dept_clean[["Total_hogares_miles"]]
y = df_dept_clean["Hogares_internet_miles"]

# Entrenamiento del modelo de Regresión Lineal
modelo_lr = LinearRegression()
modelo_lr.fit(X, y)

# Predicciones y métricas de evaluación
y_pred = modelo_lr.predict(X)
pendiente = modelo_lr.coef_[0]
intercepto = modelo_lr.intercept_
r2 = r2_score(y, y_pred)
rmse = np.sqrt(mean_squared_error(y, y_pred))

print("=== PARÁMETROS DEL MODELO DE REGRESIÓN LINEAL ===")
print(f"Pendiente (m / beta_1): {pendiente:.4f}")
print(f"Intercepto (b / beta_0): {intercepto:.4f}")
print(f"Ecuación: y = {pendiente:.4f} * X + ({intercepto:.4f})")
print(f"Coeficiente de Determinación (R^2): {r2:.4f}")
print(f"Error Cuadrático Medio (RMSE): {rmse:.4f} mil hogares")
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from sklearn.cluster import KMeans

# Definición de variables para la clusterización
X_cluster = df_dept_clean[['Total_hogares_miles', 'Porcentaje_internet']]

# Configuración y entrenamiento de K-Means con K=3
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df_dept_clean['Cluster_Id'] = kmeans.fit_predict(X_cluster)

# Mapeo y caracterización de los Clústeres ordenados por volumen poblacional
resumen_clusters = df_dept_clean.groupby('Cluster_Id').agg({
    'Total_hogares_miles': 'mean',
    'Hogares_internet_miles': 'mean',
    'Porcentaje_internet': 'mean',
    'Departamento': 'count'
}).rename(columns={'Departamento': 'Cant_Departamentos'}).reset_index()

print("=== CENTROIDES Y CARACTERIZACIÓN DE CLÚSTERES ===")
print(resumen_clusters)
```
